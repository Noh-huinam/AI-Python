```python
#asmd함수정의
def asmd(x, y, op):
    if op == '+':
        print(f"{x} + {y} = {x+y}")
    elif op == '-':
        print(f"{x} - {y} = {x-y}")
    elif op == '*':
        print(f"{x} * {y} = {x*y}")
    elif op == '/':
            print(f"{x} / {y} = {x/y}")
    else:
        print("지원하지 않는 연산자입니다.")
```

```python
x = int(input('첫번째 수:'))
y = int(input('두번째 수:'))
op=input('연산자를 입력하시오')
asmd(x,y,op)
```
