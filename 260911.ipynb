### 변수 사용 예시

```python
# 1. 변수 선언 및 값 할당
my_string = "Hello, World!" # 문자열 할당
my_integer = 100            # 정수 할당
my_float = 3.14             # 실수 할당
is_active = True            # 불리언(참/거짓) 할당
my_list = [1, 2, 3]         # 리스트 할당

# 2. 변수의 값 출력
print("문자열 변수:", my_string)
print("정수 변수:", my_integer)
print("실수 변수:", my_float)
print("불리언 변수:", is_active)
print("리스트 변수:", my_list)

# 3. 변수의 값 변경
my_integer = 200 # my_integer의 값을 100에서 200으로 변경
print("변경된 정수 변수:", my_integer)

# 4. 다른 변수를 이용한 연산
result = my_integer + 50
print("연산 결과:", result)
```

```python
#두 자연수 10와 5가 주어진다.
#이때, A+B, A-B, A*B, A/B(몫), A%B(나머지)를 출력하는 프로그램을 작성하시오

A=10
B=5

print("더하기 값:", A+B)
print("빼기 값:", A-B)
print("곱한 값:", A*B)
print("나눈 몫:", A/B)
print("나눈 나머지:", A%B)
```

```python
#가로 12, 세로 7인 직사각형의 넓이와 둘레를 구하시오

A=12
B=7
print("직사각형의 넓이:", A*B)
print("직사각형의 둘레:", 2*(A+B))
```

```python
#국어 88, 영어 92, 수학75점인 학생의 과목 평균을 구하시오

A=88
B=92
C=75
print("평균:", (A+B+C)/3)
```
