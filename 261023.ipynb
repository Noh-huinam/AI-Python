```python
#과제 1. 학생 세 명('민수':85, '지영':92, '현우':78)의 점수를 담은 딕셔너리 scores를 만드세요. 이어서 '서연':88을 추가하고, '현우'의 점수를 재시험 결과인 83으로 수정한 뒤 딕셔너리와 항목 개수를 출력하세요.

scores = {'민수': 85, '지영': 92, '현우': 78}

scores['서연'] = 88  
scores['현우'] = 83  

print(scores)
print(len(scores))
```

```python
#사용 가능한 프로그래밍 언어 조사

langs = {}

while True:
    s = input('사용 가능한 언어(,로 구분) 또는 종료: ')
    s = s.lower()
    if s=='종료':
        print('종료합니다.')
        break
    for a in s.split(','):
        if a in langs:
            langs[a] += 1
        else:
            langs[a] = 1

for key, value in langs.items():
    print(key, value)
```

```python
#수도명을 보고 국가를 맞추는 프로그램

import random

capitals = {
    "대한민국": "서울",
    "일본": "도쿄",
    "중국": "베이징",
    "미국": "워싱턴 D.C.",
    "캐나다": "오타와",
    "영국": "런던",
    "프랑스": "파리",
    "독일": "베를린",
    "이탈리아": "로마",
    "스페인": "마드리드",
    "호주": "캔버라",
    "뉴질랜드": "웰링턴",
    "인도": "뉴델리",
    "태국": "방콕",
    "베트남": "하노이",
    "브라질": "브라질리아",
    "아르헨티나": "부에노스아이레스",
    "이집트": "카이로",
    "케냐": "나이로비",
    "튀르키예": "앙카라",
}

while True:
    nation, capital = random.choice(list(capitals.items()))

    print(f"\n수도: {capital}")
    answer = input('이 수도의 국가명은? (끝내려면 "종료" 입력): ').strip()

    if answer == "종료":
        print("프로그램을 종료합니다.")
        break

    elif answer == nation:
        print("정답입니다!")
    else:
        print(f"틀렸습니다. 정답은 {nation}입니다.")
```
