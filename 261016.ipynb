```python
# 1부터 100까지 합을 출력하시오

sum=0
for i in range(1,101):
    sum=sum+i
print(sum)
```

```python
# 1부터 100까지 짝수의 합(if문 사용)

sum=0
for i in range(1,101):
    if i%2==0:
        sum+=i

print(sum)
```

```python
# 1부터 100까지 짝수의 합(if문 사용x)

sum=0
for i in range(2,101,2):
    sum+=i
print(sum)
```

```python
# 정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요.

n=int(input("정수를 입력하세요:"))
for i in range(1,10):
  print(n,"x",i,"=",n*i)
```

```python
# 2단부터 9단까지 구구단을 7x1=7 형식으로 출력하세요.

for i in range(2,10):
  print("-"*10)
  for j in range(1,10):
        print(i, "X", j, "=", i*j)
```

```python
# while문으로 1-100까지 합을 출력하시오.

sum=0
i=1
while i<=100 :
  sum+=i
  i+=1
print(sum)
```

```python
sum=0
i=1
while True:
  sum+=i
  i+=1
  if i>100:
    break
print(sum)
```

```python
# 양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요. (while True와 break 사용)

sum=0
while True:
  i=int(input("양의 정수를 입력하시오:"))
  if i==0:
    break
  sum+=i
print(sum)
```

```python
# 1부터 20까지의 수 중 짝수만 담은 리스트를 리스트로 만들어 출력하세요.

evens=[]
for i in range(1,21):
    if i%2==0:
      evens.append(i)
print(evens)


# 16 제거 해보기 (pop())
evens.pop(7)
print(evens)

# 12 제거 (del)
del evens[-4]
print(evens)

# 10 제거 (remove 삭제할 값 직접입력)
evens.remove(10)
print(evens)


print(evens[5])
print(evens[-2])
```

```python
# sum, max 사용하는 경우
scores = [85, 95, 70, 75, 99, 93, 88]
print('합계:', sum(scores))
print('최댓값:',max(scores))
```

```python
# sum 사용하지 않는 경우

total=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
    total+=i
print(total)
```

```python
# max를 사용하지 않는 경우
max_score=0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
    if max_score < i:
        max_score=i
print(max_score)
```
